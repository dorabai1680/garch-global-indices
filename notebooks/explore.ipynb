```python
import sys
from pathlib import Path

ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
sys.path.insert(0, str(ROOT))

import config
from src.data_loader import load_dataset
from src.analysis import fit_garch, stationarity_table, summary_stats

prices, log_ret, _ = load_dataset(
    config.INDICES,
    start=config.START_DATE,
    cache_path=ROOT / "data" / "global_indices_adj_close.csv",
    return_scale=config.RETURN_SCALE,
)
log_ret = log_ret.dropna(how="any")
log_ret.tail()
```

```python
summary_stats(log_ret).round(4)
```

```python
stationarity_table(log_ret).round(4)
```

```python
fit = fit_garch(log_ret["S&P 500"], p=1, q=1, dist="normal")
print(f"ω={fit.omega:.6f}, α={fit.alphas[0]:.4f}, β={fit.betas[0]:.4f}, α+β={fit.persistence:.4f}")
fit.conditional_vol.plot(figsize=(12, 4), title="S&P 500 conditional volatility")
```
